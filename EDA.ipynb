```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Tugas Analisa 2 :
jawaban : 1. Kolom yang Memiliki Data Kosong (None)
Kolom yang terlihat memiliki data kosong atau nilai None adalah kolom 'menu' dan kolom 'terjual'.
• Kolom 'menu': Memiliki nilai None pada baris terakhir (indeks ke-4).
• Kolom 'terjual': Memiliki nilai None pada baris ketiga (indeks ke-2).


```python
df.to_csv('data_kantin.csv', index=False) # Menyimpan DataFrame ke file CSV terlebih dahulu
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Tugas Analisa 3 :
jawaban : Karena gambar tersebut tidak menampilkan output visual atau teks dari eksekusi kode df.info(), kolom spesifiknya tidak dapat ditunjuk secara langsung. Namun, secara konsep pemrograman Python (Pandas),

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisa 4 :
jawaban : . Mengapa Kolom 'terjual' Diisi dengan Angka 0 (fillna(0))?
Asumsi Logis Konstanta: Jika data penjualan suatu produk kosong (NaN/null), asumsi yang paling masuk akal dalam konteks bisnis adalah produk tersebut tidak ada yang membeli atau belum laku sama sekali.
Mempertahankan Data: Mengisi dengan angka 0 membuat kita tetap bisa mempertahankan baris data (informasi produk) tersebut untuk analisis lainnya, tanpa merusak perhitungan statistik seperti total penjualan atau rata-rata penjualan.


```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Tugas Analisa 5 :
jawaban : Jumlah baris dataset setelah perintah drop_duplicates() akan berkurang sebanyak jumlah baris duplikat yang ditemukan sebelumnya. Secara matematis, rumusnya adalah:
\(\text{Jumlah\ Baris\ Baru}=\text{Jumlah\ Baris\ Awal}-\text{Jumlah\ Baris\ Duplikat}\)

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Tugas Analisa 6 :
jawaban : 1. Menu dengan total_pendapatan Tertinggi
Kode tersebut melakukan pengelompokan data berdasarkan kolom makanan/minuman ('menu'), lalu menjumlahkan (.sum()) kolom 'total_pendapatan' untuk setiap menu tersebut.
Secara teoritis, hasil dari sintaks print(ringkasan) akan menampilkan daftar menu beserta total pendapatannya. Menu yang menghasilkan angka total_pendapatan tertinggi adalah menu yang berada di posisi paling atas jika diurutkan menurun, atau memiliki nominal terbesar dalam deretan data agregasi tersebut.
